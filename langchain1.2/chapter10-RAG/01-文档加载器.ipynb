```python

# 1、加载txt文件
from langchain_community.document_loaders import TextLoader
from rich import print as rprint

loader = TextLoader(
    "../asset/load/01-langchain-utf-8.txt",
    encoding="utf-8",
)
docs = loader.load()

rprint(docs)

loader = TextLoader(
    "../asset/load/01-langchain-gbk.txt",
    encoding="gbk",
)
docs = loader.load()

rprint(docs)

```

```python
# 加载csv
from langchain_community.document_loaders import CSVLoader

loader = CSVLoader(
    "../asset/load/02-load.csv",
)
docs = loader.load()
rprint(docs)
```

```python
from langchain_community.document_loaders import JSONLoader

json_loader = JSONLoader(
    file_path="../asset/load/03-load.json", 
    jq_schema=".", # 直接提取完整的JSON对象
    text_content=False, # 保持原始JSON结构，将提取的数据转换为JSON字符串存入page_content字段中
)

docs = json_loader.load()
rprint(docs)
```
