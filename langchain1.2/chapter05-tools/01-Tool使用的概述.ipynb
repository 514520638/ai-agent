```python
# 见  工具调用流程.png

# Tool使用的概述
from langchain_core.tools import tool;
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv;
import os;
from rich import print as rprint

load_dotenv(override=True)  # 覆盖环境变量

model = ChatOpenAI(
    model="MiniMax-M3",
    api_key=os.getenv("api_key"),
    base_url=os.getenv("base_url"),
)
```

```python
# 使用tool调用
@tool
def get_weather(city: str) -> str:
    """
    获取指定城市的天气信息
    """
    # 这里可以调用实际的天气API获取天气信息
    return f"{city}的天气是晴天，温度25摄氏度。"

get_weather.invoke({"city": "北京"})
```

```python
# 基于模型调用
model.invoke("请介绍你自己,你是什么大模型")

# 绑定工具
model_with_tools = model.bind_tools([get_weather])

# response = model_with_tools.invoke("请帮我查询北京的天气")

# 结合message调用
message=[
    {"role": "user", "content": "请帮我查询广东的天气"}
]
response = model_with_tools.invoke(message)

rprint(response)
```
