```python
from langchain_openai import ChatOpenAI;
from dotenv import load_dotenv;
import os;
from rich import print as rprint;
from langchain_core.utils.function_calling import convert_to_openai_tool;
from langchain_core.tools import tool;

load_dotenv(override=True)  # 覆盖环境变量

model = ChatOpenAI(
    model="MiniMax-M3",
    api_key=os.getenv("api_key"),
    base_url=os.getenv("base_url"),
)
@tool(parse_docstring=True)
def get_weather(city: str= "北京") -> str:
    """
    获取指定城市的天气信息。

    Args:
        city (str): 城市名称，默认为北京。
    """
    return f"{city}的天气是晴天，温度25摄氏度。"

model_with_tools = model.bind_tools([get_weather])  

# convert_to_oepnai_tool将传入函数转换成Tool实例
rprint(convert_to_openai_tool(get_weather))

response = model_with_tools.invoke("请告诉我北京的天气情况")
rprint(response)
```

```python


@tool(description="获取指定城市的地理位置", )
def get_location(city: str= "上海") -> str:
    return f"{city}城市在中国的东部地区，位于长江入海口。"

model_with_tools = model.bind_tools([get_weather, get_location])    
response2=model_with_tools.invoke("请告诉我北京的地理位置")
rprint(convert_to_openai_tool(get_location))
rprint(response2)
```
