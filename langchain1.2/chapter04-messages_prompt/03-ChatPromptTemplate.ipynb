```python
from langchain_core.prompts import ChatPromptTemplate;

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个非常友好的AI助手，你的名字是{name}"),
    ("human", "你好，你最近怎么样？"),
    ("ai", "我很好，谢谢！你呢？"),
    ("human", "{user_input}"),
])

# ChatPromptTemplate 使用invoke调用
result = chat_prompt_template.invoke({"name": "小磊", "user_input": "我叫什么名字？"})
# 相当于生成了新的AI提示词
print(result.messages)

```
